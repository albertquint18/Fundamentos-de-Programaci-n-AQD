### Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión y, al final, imprime con `items()` el color ganador junto con su total.

```python
votos = {
    "Rojo": 0,
    "Azul": 0,
    "Verde": 0,
}
for v in range(5):
    votcont = input(f"Ingresa por quien quieras votar {v + 1} (Rojo, Azul, Verde)")
    votos[votcont] += 1
for total in votos.items():
    print (total,)
```

### Desafío Extra 3: Validación robusta de entrada
Escribe un programa que pida al usuario dos números enteros dentro de un bloque `try-except`. Si el usuario ingresa texto o un valor vacío, muestra un mensaje controlado y repite la solicitud con un ciclo `while`. Luego pide la división del primero entre el segundo, con un bloque específico para `ZeroDivisionError`.

```python
while True:
    try:
        numero1 = int(input("Dame un número: "))
        numero2 = int(input("Ingresa el otro numero"))
        break
    except ValueError:
        print("Error, ingresa dato valido")
try:
    resultado = numero1 / numero2
except ZeroDivisionError:
    print("Error de division entre cero")
else:
    print(f"El resultado es: {resultado}")
finally:
    print("Fue un gusto ayudarte")
    
```

### Desafío Extra 4: Analizador de textos
Crea un programa que reciba una frase, imprima su longitud con `len()`, la convierta a mayúsculas, reemplace la palabra "Python" por "programación" y muestre cuántas palabras contiene usando una función que retorne el conteo.

```python
frase = input("Ingresa la frase con Python: ")
print(frase)
print(f"La longitud es: {len(frase)}")
print(f"Mayuscula: {frase.upper()}")
print(f"Remplazo: {frase.replace("Python", "programación")}")
def contador(frase):
    pal = frase.split()
    return len(pal)
print(f"la frase tiene estas palabras: {contador(frase)}")
```

### Desafío Extra 5: Minimenú modular integrador
Construye un menú con un ciclo `while` y opciones numéricas que permita ejecutar de forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en el diccionario `contactos`, (3) dividir dos números con manejo de `ZeroDivisionError` y (4) analizar un mensaje de texto. La opción (5) debe finalizar el programa.

```python
numeros = (4, 10, 5, 9, 3)
def ordenar_num():
    print(sorted(numeros))
contactos = {
    "Paulinho": "4421236767",
    "Gus": "4423459024",
    "Carlitos": "4426967345"
}
def buscar_phon():
    nom = input("Ingresa el contacto a buscar: ")
    if nom in contactos:
        print(contactos[nom])
    else:
        print("Ese bro no esta aquí")
def dividir_n():
    try:
        num1 = int(input("Ingresa el primer número"))
        num2 = int(input("Ingresa el segundo número"))
        resultado = num1 / num2
        print(f"El resultado es: {resultado}")
    except ZeroDivisionError:
        print("Error, no se puede")
def analiz_men():
    mens = input("Esribe un mensaje con futuro: ")
    print(f"longitud: {len(mens)}")
    print(f"Mayuscula: {mens.upper()}")
    print(f"Remplazo con pasado: {mens.replace("futuro", "pasado")}")
while True:
    print("------MENÚ------")
    print("1- mostrar la tupla de números")
    print("2- buscar un teléfono en el diccionario")
    print("3- dividir dos números")
    print("4- analizar mensajes")
    print("5- finalizar programa")
    eleccion = input("Elige la opción deseada: ")
    if eleccion == "1":
        ordenar_num()
    elif eleccion == "2":
        buscar_phon()
    elif eleccion == "3":
        dividir_n()
    elif eleccion == "4":
        analiz_men()
    elif eleccion == "5":
        break
    else:
        print("opción invalida")
```
