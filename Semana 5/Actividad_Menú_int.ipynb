## 1. Descripción del reto

Se creo una **aplicación interactiva en Python** que demuestra el uso integral de **tuplas**, **diccionarios**, **manipulación de cadenas de texto (strings)** y **manejo robusto de excepciones**, estructurada a través de un **menú principal modular** controlado por un ciclo. El usuario puede seleccionar por número qué sección desea ejecutar (Tuplas, Diccionarios, Excepciones, Strings o Finalizar) y el programa delega cada tarea a funciones modulares.

Este reto integra la creación y manipulación de **tuplas** (Tema 13), los **diccionarios** con claves y valores (Tema 14), los bloques **try-except** con excepciones específicas (Tema 15) y los **métodos de strings** con conteo de palabras (Tema 16), además de consolidar las funciones con y sin retorno vistas en semanas anteriores.

#### Código:

```python
numeros = (5, 15, 10, 20, 25)
def suma_tu(tupla):
    suma = sum(tupla)
    return suma
def tupls_nu():
    print("La tupla es: ", numeros)
    print("Tercer elemento:", numeros[2])
    nuevnu1 = int(input("Ingresa un número nuevo"))
    nuevnu2 = int(input("Ingresa el segundo número nuevo"))
    nuevtup = (nuevnu1, nuevnu2, 5, 15, 10, 20, 25)
    lista_nums = list(nuevtup)
    lista_nums.sort()
    print("Lista ordenada: ", lista_nums)
    total = suma_tu(nuevtup)
    print("Suma final:", total)
contactos = {
    "Emi": "4428906756",
    "Mario": "4423459872",
    "Uri": "4421254578"
}
def busc_phn(contactos, nombre):
    if nombre in contactos:
        return contactos[nombre]
    else:
        return "No se encontro"
def diccionar_cont():
    nuev_nom = input("Ingresa el nuevo nombre: ")
    nuev_tel = int(input("Ingresa el nuevo número: "))
    contactos[nuev_nom] = nuev_tel
    print("Contactos")
    for nombre in contactos:
        print(nombre)
    buscon = input("A quien buscas?")
    telef = busc_phn(contactos, buscon)
    print("Telefono:", telef)
def exep_us():
    try:
        nom1 = int(input("Ingresa un número: "))
        nom2 = int(input("Ingresa el segundo número: "))
        resultadosum = nom1 + nom2
        print("El resultado de la suma es: ", resultadosum)
        try:
            resultadodiv = nom1 / nom2
            print("La división es:", resultadodiv)
        except ZeroDivisionError:
            print("Error, no se puede dividir entre 0")
    except ValueError:
        print("Error, los datos están mal")
def string_us():
    mensaje = input("Ingresa un mensaje con la palabra: spiderman")
    print(mensaje)
    print(f"La longitud es: {len(mensaje)}")
    print(f"Mensaje en mayusculas: {mensaje.upper()}")
    print(f"Frase con remplazo de palabra: {mensaje.replace("spiderman", "venom")}")
    def cantidad(mensaje):
        mortygucci = mensaje.split()
        return len(mortygucci)
    print(f"La frase tiene esta cantidad de plabras: {cantidad(mensaje)}")
while True:
    print("------MENÚ------")
    print("1- Tuplas")
    print("2- Diccionarios")
    print("3- Excepciones")
    print("4- Strings")
    print("5- finalizar programa")
    eleccion = input("Elige la opción deseada: ")
    if eleccion == "1":
        tupls_nu()
    elif eleccion == "2":
        diccionar_cont()
    elif eleccion == "3":
        exep_us()
    elif eleccion == "4":
        string_us()
    elif eleccion == "5":
        break
    else:
        print("opción invalida")
```
