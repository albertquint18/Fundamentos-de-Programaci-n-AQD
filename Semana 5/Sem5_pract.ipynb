```python
def suma_tupla(t):
    total = 0
    for i in t:
        total += i
    return total
calorias = (800, 950, 700, 1300, 800,)
resultado = suma_tupla(calorias)
print(f"El total de calorias es: {resultado}")
```

### Diccionarios 

```python
contactos = {
    "Ricardo": "4455667788",
    "Juan": "3344556677",
    "Luisa": "6677334455"
}
contactos["Cris"] = "2990556677"
contactos["Ricardo"] = "3344556677"

print(contactos)
```

```python
print(contactos.keys())
print(contactos.values())
print(contactos.items())
```

```python
for n in contactos.keys():
    print(n)
```

```python
print("Impresion de llaves")
for n in contactos.keys():
    print(n)

print("Impresion de valores")
for v in contactos.values():
    print(v)

print("Impresion completa")
for n, v in contactos.items():
    print(f"Nombre: {n} - Telefono: {v}")
```

```python
def buscar_contacto(contactos, nombre):
    return contactos.get(nombre, "No encontrado")

nombre = input("Ingrese el nombre del contacto a buscar: ")
telefono = buscar_contacto(contactos, nombre)
print(f"Telefono de {nombre}: {telefono}")
```

## Excepciones

```python
try:
    edad = int(input("Ingresa edad: "))
    print(f"Edad ingresada: {edad}")
except:
    print("Error")
```

```python
try:
    dividiendo = int(input("dame un numero: "))
    divisor = int(input("Entre que numero lo dividimos: "))
    resultado = dividiendo / divisor
except ZeroDivisionError:
    print("Error de division entre cero")
else:
    print(f"El resultado es: {resultado}")
finally:
    print("Fue un gusto ayudarte")
```
