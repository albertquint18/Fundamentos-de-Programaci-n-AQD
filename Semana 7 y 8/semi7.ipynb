```python
with open("bitacora.txt", mode="w") as archivo:
    archivo.write("Esto es una primera linea\n")
    archivo.write("Esto es una segunda linea\n")
    archivo.write("Esto es una tercera linea\n")

with open("bitacora.txt", mode="a") as archivo:
    archivo.write("Esto es una cuarta linea\n")
```
