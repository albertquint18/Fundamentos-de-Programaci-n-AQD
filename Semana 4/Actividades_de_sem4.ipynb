```python
# Preguntar cuantas calificaciones desea promediar, y promediarlas
def promedio(sumatoria, n):
    return sumatoria / n
n = int(input("Cuántas calificaciones quieres promediar?"))
sumatoria = 0

print(f"Se van a pedir {n} calificaciones")
for c in range(1, n+1):
    calif = float(input(f"Dame la calificación {c}"))
    sumatoria = sumatoria + calif
    print(f"Hasta este punto, la sumatoria va en {sumatoria}")
```
