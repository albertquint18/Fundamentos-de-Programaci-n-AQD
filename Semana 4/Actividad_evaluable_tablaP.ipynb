##### Descripción:
Se Desarrollo una **aplicación en Python** que muestra una **tabla de Pitágoras interactiva** para facilitar la multiplicación de dos factores ingresados por el usuario. El programa almacena la tabla en una **lista de listas (matriz)**, la imprime y permite que el usuario capture dos factores (renglón y columna) para obtener su producto consultando directamente la información almacenada en la matriz.
Se integraron listas, matrices, funciones que no regresan valor y funciones que si regresan valor

##### Requerimientos técnicos obligatorios

1. **Matriz obligatoria en lista de listas:** La tabla de Pitágoras debe almacenarse obligatoriamente en una **lista de listas (matriz)**, construida con ciclos `for` y `append()`.
2. **Función que no regresa valor para imprimir la matriz:** El programa debe generar e imprimir en pantalla la matriz completa utilizando una **función que no regrese valor** (sin el uso de la palabra clave `return`).
3. **Visualización sin corchetes ni comas:** La salida en pantalla debe mostrar los números tabulados **sin corchetes ni comas** que evidencien la sintaxis de lista interna de Python.
4. **Captura de dos factores:** El usuario debe capturar dos factores numéricos (renglón y columna) para realizar una multiplicación.
5. **Función que regresa valor para multiplicar:** La multiplicación de los dos factores debe realizarse mediante una **función que regresa valor** (con la palabra clave `return`), utilizando la información almacenada en la lista de listas.
6. **Restricción crítica:** Se prohíbe terminantemente el uso de cualquier operador matemático de multiplicación (asterisco `*`) en la función que calcula el resultado. El resultado se debe obtener **consultando la posición de los factores en la matriz de Pitágoras**.


```python
print("Tabla Pitagorica")
matriz = []
for f in range(0, 11):
   fila = []
   for c in range(0, 11):
       if f == 0:
           fila.append(c)
       elif c == 0:
           fila.append(f)
       else:
           fila.append(f * c)
#Para que la matriz tenga encabezados se pregunta si f es igual a 0, si si, se agrega el valor de c, al igual que si c es igual a 0, se agrega el valor de f, y si no, se agrega el valor de f*c.
   matriz.append(fila)
def imprimir_matriz(matriz):
    for f in matriz:
        for c in f:
            print(f"{c:>4}", end=" ")
        print()
imprimir_matriz(matriz)
#El def imprimir_matriz toma la mtriz y la imprime con formato de 4 espacios, un espacio entre cada valor y al final del renglon hace un salto de linea.
def multiplicar(matriz, renglon, columna):
    return matriz[renglon][columna]
#Este def toma la matriz, el renglon y la columna, y devuleve el valor de la posición especificada, guiandose por cordendas de la matriz
renglon = int(input("Ingrese el número de renglón: "))
columna = int(input("Ingrese el número de columna: "))
resultado = multiplicar(matriz, renglon, columna)
print(f"El resultado de la multiplicación es: {resultado}")
```
