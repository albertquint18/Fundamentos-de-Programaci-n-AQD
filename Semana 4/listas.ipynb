```python
filas = 3
columnas = 3
matriz = []
for f in range(filas):
    fila_actual = []
    for c in range(columnas):
        valor = 0
        if f == c:
            valor = 1
        fila_actual.append(valor)
    matriz.append(fila_actual)
print("Impresion de matriz")
for f in range(filas):
    for c in range(columnas):
        print(matriz[f][c], end=" ")
    print()
```

```python
lista = []
contador = 1
while contador <= 10:
    valor = int(input("Introduce un número del -10 al 10: "))
    if valor >= -10 and valor <= 10:
        lista.append(valor)
        contador += 1
    else:
        print("Valor incorrecto")
print(lista)
for i in range(len(lista)):
    if i < 0:
        print("Asteriscos Negativos")
    elif i > 0:
        for j in range(10):
            print("*", end=" ")
        print()
    else:
        print("/")
```
