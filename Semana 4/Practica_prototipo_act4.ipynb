```python
print("Tabla Pitagorica")
matriz = []
for f in range(0, 11):
   fila = []
   for c in range(0, 11):
       if f == 0:
           fila.append(c)
       elif c == 0:
           fila.append(f)
       else:
           fila.append(f * c)
#Para que la matriz tenga encabezados se pregunta si f es igual a 0, si si, se agrega el valor de c, al igual que si c es igual a 0, se agrega el valor de f, y si no, se agrega el valor de f*c.
   matriz.append(fila)
def imprimir_matriz(matriz):
    for f in matriz:
        for c in f:
            print(f"{c:>4}", end=" ")
        print()
imprimir_matriz(matriz)
#El def imprimir_matriz toma la mtriz y la imprime con formato de 4 espacios, un espacio entre cada valor y al final del renglon hace un salto de linea.
def multiplicar(matriz, renglon, columna):
    return matriz[renglon][columna]
#Este def toma la matriz, el renglon y la columna, y devuleve el valor de la posición especificada, guiandose por cordendas de la matriz
renglon = int(input("Ingrese el número de renglón: "))
columna = int(input("Ingrese el número de columna: "))
resultado = multiplicar(matriz, renglon, columna)
print(f"El resultado de la multiplicación es: {resultado}")
```
