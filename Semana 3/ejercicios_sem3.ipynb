#### Ejercicios Variados

```python
numero = int(input("Ingresa un número entero"))
if numero % 2 == 0:
    print("Es par")
else:
    print("Es impar")
```

```python
sum = 0
while True:
    num = (float(input("Ingresa números positivos")))
    sum = sum + num
    print(f"La suma es: {sum}")
    if sum >= 300:
        break
   
```

```python
num = int(input("Ingresa el número"))
invertido = 0
while num != 0:
    r = num % 10
    invertido = invertido * 10 + r
    num = num // 10
print(f"{invertido}")
```
